# 简单查看频谱对比

```python

import numpy as np
import matplotlib.pyplot as plt

# 1. 加载预训练数据
pre_path = "SaveData/PreTrain/D1-PreTrain.npy"
pre_data = np.load(pre_path, allow_pickle=True).item()
pre_sample = pre_data["dataset"][0]  # (2, 1000)

# 2. 加载微调数据
fine_path = "SaveData/FineTune/D1-Undefence-FineTune.npz"
with np.load(fine_path) as fine_data:
    fine_sample = fine_data["X_tune_train"][0]  # (2, 1000)

# 3. 直接计算 FFT 幅度谱，不减去均值
def get_spectrum(signal):
    n = len(signal)
    fft_result = np.fft.rfft(signal)
    magnitude = np.abs(fft_result) / n

    # 转换为单边幅度谱
    if n % 2 == 0:
        magnitude[1:-1] *= 2
    else:
        magnitude[1:] *= 2

    freq = np.fft.rfftfreq(n, d=1.0)
    return freq, magnitude

# 4. 绘制 4 幅频谱图
fig, axes = plt.subplots(2, 2, figsize=(14, 8))

samples = [
    (pre_sample, "PreTrain - Sample 0", 0, 0),
    (pre_sample, "PreTrain - Sample 0", 1, 1),
    (fine_sample, "FineTune - Sample 0", 0, 0),
    (fine_sample, "FineTune - Sample 0", 1, 1),
]

for sample, name, ch, pos in samples:
    row = 0 if name.startswith("PreTrain") else 1
    col = ch

    freq, magnitude = get_spectrum(sample[ch])

    axes[row, col].plot(freq, magnitude)
    axes[row, col].set_title(f"{name}, Channel {ch}")
    axes[row, col].set_xlabel("Frequency (cycles/sample)")
    axes[row, col].set_ylabel("Magnitude")
    axes[row, col].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

# 预训练 / 目标域微调第一条样本：时域 -> 频域

样本是长度 1000 的均匀时间栅格上的**包计数**序列（ch0 上行、ch1 下行），
栅格宽度就是 `slot_duration` 秒，因此采样率 `fs = 1 / slot_duration`，
用 `np.fft.rfft` 得到单边频谱，频率轴 `0 ~ fs/2`。

对比的两条：

| | 文件 | 来源 trace |
|---|---|---|
| 源域·预训练 | `SaveData/PreTrain/D1-PreTrain.npy` | `Traces/D1-Undefence/0-0` |
| 目标域·微调 | `SaveData/FineTune/D2-Undefence-FineTune.npz` | `Traces/D2-Undefence/0-0` |

**注意 `slot_duration` 不能混用**：`CIF.fun()` 里 `slot_duration = 3 * load_time / 1000` 并截断到 `[0.02, 0.08]`，
是**按每条 trace 自适应**的。实测第一条：D1 = `0.034756s`（奈奎斯特 14.39Hz），D2 = `0.052088s`（奈奎斯特 9.60Hz）。
微调的 `.npz` 里**没有存** `slot_duration`，所以下面从原始 trace 回算，直接拿 D1 的值会把 D2 的频率轴画歪。

```python
# 将预训练(源域 D1)第一条样本与微调(目标域 D2)第一条样本转换到频域
# 本 cell 自包含：前面的 cell 没跑过也能单独运行（变量不存在时自动重新加载）
# 图的文字一律用英文：matplotlib 默认字体不含中文字形，写了中文会变成空框
import sys
import numpy as np
import matplotlib.pyplot as plt
from CIF import fun

plt.rcParams["axes.unicode_minus"] = False

PRE_NPY = "SaveData/PreTrain/D1-PreTrain.npy"
FT_NPZ = "SaveData/FineTune/D2-Undefence-FineTune.npz"
FT_TRACE = "Traces/D2-Undefence/0-0"      # 对应 X_tune_train[0] 的原始 trace


def slot_duration_from_trace(path):
    """按 ProcessData.py 的同一套流程，从原始 trace 重算 load_time / slot_duration"""
    times, sizes = [], []
    with open(path) as f:
        for line in f:
            t, s = line.rstrip("\n").split("\t")[:2]
            times.append(float(t))
            sizes.append(int(float(s)))
    times = np.asarray(times) - float(times[0])
    _, load_time, slot = fun(times, np.asarray(sizes))
    return load_time, slot


try:
    dataset, label, slot_duration
except NameError:
    _pre = np.load(PRE_NPY, allow_pickle=True).item()
    dataset, label, slot_duration = _pre["dataset"], _pre["label"], _pre["slot_duration"]

try:
    X_train, y_train
except NameError:
    with np.load(FT_NPZ) as _ft:
        X_train, y_train = _ft["X_tune_train"], _ft["Y_tune_train"]

N = dataset.shape[-1]

# 预训练第一条（源域 D1）：slot_duration 直接存在 npy 里
pre_sample, pre_label, pre_slot = dataset[0], label[0], float(slot_duration[0])

# 微调第一条（目标域 D2）：npz 没存 slot_duration，而 D2 每条 trace 的 slot 都不同，
# 必须回原始 trace 重算，否则频率轴刻度是错的
try:
    _, ft_slot = slot_duration_from_trace(FT_TRACE)
    print(f"微调 slot_duration 由 {FT_TRACE} 重算得到: {ft_slot:.6f}s")
except FileNotFoundError:
    ft_slot = pre_slot
    print(f"警告: 找不到 {FT_TRACE}，微调暂用预训练的 slot，频率轴不准确")

ft_sample, ft_label = X_train[0], y_train[0]
print("预训练第一条 == 微调第一条:", np.array_equal(pre_sample, ft_sample))

# 供下一个 cell（叠加对比图）使用
samples = [
    ("Pretrain D1 (source)", pre_sample, pre_label, pre_slot),
    ("Finetune D2 (target)", ft_sample, ft_label, ft_slot),
]

fig, axes = plt.subplots(len(samples), 3, figsize=(16, 7))

for row, (name, sample, lab, slot) in enumerate(samples):
    x = sample.astype(np.float64)              # (2, 1000)
    fs = 1.0 / slot                            # sampling rate (Hz)
    freqs = np.fft.rfftfreq(N, d=slot)         # 0 ~ fs/2
    amp = np.abs(np.fft.rfft(x, axis=1)) / N   # single-sided magnitude
    db = 20 * np.log10(amp + 1e-12)            # magnitude in dB
    t = np.arange(N) * slot                    # time axis (s)

    for ch, (ch_name, color) in enumerate([("ch0 upstream", "tab:blue"),
                                           ("ch1 downstream", "tab:red")]):
        axes[row, 0].plot(t, x[ch], color=color, lw=0.8, label=ch_name)
        axes[row, 1].plot(freqs, amp[ch], color=color, lw=0.8, label=ch_name)
        axes[row, 2].plot(freqs, db[ch], color=color, lw=0.8, label=ch_name)

    for col, (ttl, xl, yl) in enumerate([
        ("time domain (packet count)", "Time (s)", "Packets"),
        ("magnitude spectrum", "Frequency (Hz)", "Magnitude"),
        ("magnitude spectrum (dB)", "Frequency (Hz)", "Magnitude (dB)"),
    ]):
        ax = axes[row, col]
        ax.set_title(f"{name} | label={lab} | {ttl}")
        ax.set_xlabel(xl)
        ax.set_ylabel(yl)
        ax.grid(True, alpha=0.3)
        ax.legend(fontsize=8)

    print(f"{name}: label={lab}, slot_duration={slot:.6f}s, fs={fs:.2f}Hz, 奈奎斯特={fs/2:.2f}Hz, "
          f"包总数 ch0/ch1={x.sum(axis=1).astype(int)}")

plt.tight_layout()
plt.show()
```

```python
# 最后一张：把源域(预训练 D1)与目标域(微调 D2)第一条样本的频谱叠在一起对比
# 完全自包含：单独运行本 cell 也能出图，不依赖前面任何 cell
if "samples" not in globals():
    import numpy as np
    from CIF import fun

    PRE_NPY = "SaveData/PreTrain/D1-PreTrain.npy"
    FT_NPZ = "SaveData/FineTune/D2-Undefence-FineTune.npz"
    FT_TRACE = "Traces/D2-Undefence/0-0"

    _pre = np.load(PRE_NPY, allow_pickle=True).item()
    dataset, label, slot_duration = _pre["dataset"], _pre["label"], _pre["slot_duration"]

    with np.load(FT_NPZ) as _ft:
        X_train, y_train = _ft["X_tune_train"], _ft["Y_tune_train"]

    _times, _sizes = [], []
    with open(FT_TRACE) as f:
        for _line in f:
            _t, _s = _line.rstrip("\n").split("\t")[:2]
            _times.append(float(_t))
            _sizes.append(int(float(_s)))
    _times = np.asarray(_times) - float(_times[0])
    _, _ft_slot = fun(_times, np.asarray(_sizes))[1:]

    samples = [
        ("Pretrain D1 (source)", dataset[0], label[0], float(slot_duration[0])),
        ("Finetune D2 (target)", X_train[0], y_train[0], _ft_slot),
    ]
    print(f"自行加载: 预训练 slot={samples[0][3]:.6f}s, 微调 slot={samples[1][3]:.6f}s")
else:
    print("复用上一个 cell 的结果")

import matplotlib.pyplot as plt

plt.rcParams["axes.unicode_minus"] = False

# 两组采样率不同(28.77Hz vs 19.20Hz)，只在共同可比较的频段里叠加
nyq = min(1.0 / s[3] for s in samples) / 2

fig, axes = plt.subplots(2, 2, figsize=(13, 7))

for ch in range(2):
    ch_name = ["ch0 upstream", "ch1 downstream"][ch]
    for row, ylab in enumerate(["Magnitude", "Magnitude (dB)"]):
        ax = axes[row, ch]
        for si, (name, sample, lab, slot) in enumerate(samples):
            x = sample.astype(np.float64)
            freqs = np.fft.rfftfreq(x.shape[-1], d=slot)
            amp = np.abs(np.fft.rfft(x, axis=1)) / x.shape[-1]
            y = amp[ch] if row == 0 else 20 * np.log10(amp[ch] + 1e-12)
            ax.plot(freqs, y, lw=0.9, ls=["-", "--"][si],
                    color=["tab:blue", "tab:orange"][si],
                    label=f"{name}  fs={1 / slot:.1f}Hz")
        ax.set_xlim(0, nyq)
        ax.set_title(f"{ch_name} | {ylab} spectrum")
        ax.set_xlabel("Frequency (Hz)")
        ax.set_ylabel(ylab)
        ax.grid(True, alpha=0.3)
        ax.legend(fontsize=8)

fig.suptitle("Sample 0 spectrum: source domain D1-PreTrain (solid) vs "
             f"target domain D2-Undefence (dashed), x limited to common 0-{nyq:.1f}Hz")
plt.tight_layout()
plt.show()
```

# D1 单条轨迹 vs D2 全部 0 类（网站 0）的分布

前两节各看了一条样本，这里换成**不对等**的画法：

| | 条数 | 画法 |
|---|---|---|
| D1 预训练 | **1 条**（`label == 0` 的第一条，`dataset[0]`） | 一条实线，只表示"这一条长什么样" |
| D2 微调（目标域） | **100 条**（`Traces/D2-Undefence/0-0 ~ 0-99`） | mean 虚线 + 两层浅色阴影 |

阴影两层，颜色都是浅色（同一色系不同透明度）：

- **外层更浅**（`alpha=0.10`）= `min ~ max` 包络，即这 100 条的**变化范围**
- **内层稍深**（`alpha=0.25`）= `mean ± std`，即典型波动范围

> 这里的 `label 0` 是**监控网站 0**（`ProcessData.py` 的编号）。开放世界的未监控类只在 `*-FineTune-open.npz` 里。

**关键处理**：100 条 D2 trace 的 `slot_duration` 逐条不同（实测 0.0343~0.0800s），频率轴也就逐条不同，**不能按同一个索引直接平均**。做法是各条先算自己的 `rfft` 幅度谱，再用 `np.interp` 插值到公共频率栅格 `0 ~ min(两边的奈奎斯特) = 6.25Hz`，然后统计 `mean / std / min / max`。D1 只有一条，直接用它自己的频率轴。

读法：把蓝色实线当成"某一条源域样本"，看它落在橙色分布带里的什么位置 —— 落在带外说明这一条比目标域的多数样本更偏离。

```python
# D1 第一条样本的频谱轨迹 vs D2 同类(网站 0)全部 100 条的分布
# 自包含，可单独运行
import numpy as np
import matplotlib.pyplot as plt
from CIF import fun

plt.rcParams["axes.unicode_minus"] = False

PRE_NPY = "SaveData/PreTrain/D1-PreTrain.npy"
D2_TRACE_DIR = "Traces/D2-Undefence"
SITE = 0
N_TRACE = 100          # 网站 0 的 trace: 0-0 ... 0-99


def trace_features(path):
    """从原始 trace 重算 CIF 特征和 slot_duration（与 ProcessData.py 同一套流程）"""
    times, sizes = [], []
    with open(path) as f:
        for line in f:
            t, s = line.rstrip("\n").split("\t")[:2]
            times.append(float(t))
            sizes.append(int(float(s)))
    times = np.asarray(times) - float(times[0])
    feat, _, slot = fun(times, np.asarray(sizes))
    return np.array(feat, dtype=np.float64), slot


# ---------- D1: 网站 0 的第一条 ----------
pre = np.load(PRE_NPY, allow_pickle=True).item()
idx0 = int(np.flatnonzero(pre["label"] == SITE)[0])
d1_x = pre["dataset"][idx0].astype(np.float64)
d1_slot = float(pre["slot_duration"][idx0])

# ---------- D2: 网站 0 全部 100 条（npz 里没存 slot_duration，回 trace 重算） ----------
d2_x, d2_slot = [], []
for i in range(N_TRACE):
    feat, slot = trace_features(f"{D2_TRACE_DIR}/{SITE}-{i}")
    d2_x.append(feat)
    d2_slot.append(slot)
d2_x, d2_slot = np.array(d2_x), np.array(d2_slot)

print(f"D1: 第 {idx0} 条 (网站{SITE} 的第一条), slot_duration={d1_slot:.6f}s")
print(f"D2: 网站{SITE} 全部 {len(d2_x)} 条, slot_duration {d2_slot.min():.4f}~{d2_slot.max():.4f}s")

# D1 只有一条，用自己的频率轴即可；D2 的 100 条频率轴逐条不同，
# 要先各自插值到公共栅格，再统计分布
nyq = min(1 / (2 * d1_slot), 1 / (2 * d2_slot.max()))
f_common = np.linspace(0, nyq, 1000)
print(f"公共频率栅格: 0 ~ {nyq:.2f} Hz")

freqs_d1 = np.fft.rfftfreq(d1_x.shape[-1], d=d1_slot)
amp_d1 = np.abs(np.fft.rfft(d1_x, axis=1)) / d1_x.shape[-1]
d1_spec = np.stack([np.interp(f_common, freqs_d1, amp_d1[ch]) for ch in range(2)])

spec2 = np.empty((len(d2_x), 2, len(f_common)))
for i, (x, slot) in enumerate(zip(d2_x, d2_slot)):
    freqs = np.fft.rfftfreq(x.shape[-1], d=slot)
    amp = np.abs(np.fft.rfft(x, axis=1)) / x.shape[-1]
    for ch in range(2):
        spec2[i, ch] = np.interp(f_common, freqs, amp[ch])

mean2, std2 = spec2.mean(axis=0), spec2.std(axis=0)
lo2, hi2 = spec2.min(axis=0), spec2.max(axis=0)

fig, axes = plt.subplots(2, 2, figsize=(13, 7))

for ch in range(2):
    ch_name = ["ch0 upstream", "ch1 downstream"][ch]
    for row, use_db in enumerate([False, True]):
        ax = axes[row, ch]
        conv = (lambda a: 20 * np.log10(np.clip(a, 1e-9, None))) if use_db else (lambda a: a)

        # 外层：D2 全部 100 条的 min~max 变化范围（更浅）
        ax.fill_between(f_common, conv(lo2[ch]), conv(hi2[ch]),
                        color="tab:orange", alpha=0.10, lw=0, label="D2 min-max (n=100)")
        # 内层：D2 的 mean +/- std（稍深）
        ax.fill_between(f_common, conv(mean2[ch] - std2[ch]), conv(mean2[ch] + std2[ch]),
                        color="tab:orange", alpha=0.25, lw=0, label="D2 mean +/- std")
        ax.plot(f_common, conv(mean2[ch]), color="tab:orange", lw=1.3, ls="--", label="D2 mean")
        ax.plot(f_common, conv(d1_spec[ch]), color="tab:blue", lw=1.6, label="D1 sample 0 (1 trace)")

        if use_db:      # dB 下 min-max 下界会很负，用 D1 曲线和 D2 mean 定 y 轴范围
            vals = np.concatenate([conv(d1_spec[ch]), conv(mean2[ch])])
            ax.set_ylim(vals.min() - 5, vals.max() + 5)

        ax.set_title(f"{ch_name} | {'Magnitude (dB)' if use_db else 'Magnitude'} spectrum")
        ax.set_xlabel("Frequency (Hz)")
        ax.set_ylabel("Magnitude (dB)" if use_db else "Magnitude")
        ax.grid(True, alpha=0.3)
        ax.legend(fontsize=8)

fig.suptitle(f"D1 single trace vs D2 site-{SITE} distribution (100 traces), common band 0-{nyq:.1f}Hz")
plt.tight_layout()
plt.show()

# D1 这一条有多大比例落在 D2 的 min-max 包络内
for ch in range(2):
    ratio = float(np.mean((d1_spec[ch] >= lo2[ch]) & (d1_spec[ch] <= hi2[ch])))
    print(f"ch{ch}: D1 单条落在 D2 min-max 包络内的频点比例 = {ratio:.1%}")
```

## 频域偏移是全局的，还是逐站点的？

前面几节（D2 / D6 / D7）都比较了 **D1 的第一条 trace vs 目标域某个网站的全部 100 条**。
单条样本的个体波动很大，看不出"这个域整体偏移了多少"。本节把口径改成**多站点**：

- 每个域取 `SITES` 个网站，每站 `N_TRACE` 条
- 先按网站求平均谱，再看**偏移在网站之间是否一致**
- 频段沿用 `_tmp_shift_metrics.py` 的 `low / mid / high` 三档，频率轴也用同一条物理轴

这一步决定了频域**到底能不能拿来跨域**：

| 观察 | 含义 | 结论 |
|---|---|---|
| 偏移的**跨站点 std** ≪ **域内 std** | 偏移是个**全局常数**，对所有网站一样 | 一次全局校正就能消掉 → 频域可以作为域不变的表示 |
| 偏移的**跨站点 std** ≳ **域内 std** | 偏移**逐网站不同**，与网站内容纠缠 | 无法全局校正 → 频域只会放大域信号，跨域更差 |

`_tmp_shift_metrics.py` 只跑了网站 0，所以给不出这个区分。下面补上。

> 图里的文字用英文，原因同前面几节：matplotlib 默认字体没有中文字形。


```python
# =====================================================================
# 准备：多站点 × 多域，并构造 4 套表示
# 下面的「分析一」「分析二」两个 cell 都依赖这里，请先运行本 cell
# =====================================================================
import numpy as np
import matplotlib.pyplot as plt
from CIF import fun

plt.rcParams["axes.unicode_minus"] = False

# ---- 配色：dataviz 参考调色板的分类槽位 1/2/3，已过 OKLab dE + CVD + 对比度校验 ----
SURFACE = "#fcfcfb"
INK     = "#0b0b0b"
INK_2   = "#52514e"
MUTED   = "#898781"
GRIDC   = "#e1e0d9"
S1, S2, S3 = "#2a78d6", "#eb6834", "#1baf7a"    # 槽位 1/2/3 -> D2 / D6 / D7

PRE_NPY = "SaveData/PreTrain/D1-PreTrain.npy"
TRACE_DIRS = {
    "D2": "Traces/D2-Undefence",   # 不同位置
    "D6": "Traces/D6-Undefence",   # 不同时间 / 内容更新
    "D7": "Traces/D7-Undefence",   # 不同浏览器版本 (TBB)
}
SITES   = list(range(10))
N_TRACE = 100
BANDS = [("low  <0.5Hz", 0.0, 0.5), ("mid 0.5-3Hz", 0.5, 3.0), ("high >3Hz", 3.0, None)]


def trace_features(path):
    # 与 ProcessData.py 同一套流程：从原始 trace 重算 CIF 和 slot_duration
    times, sizes = [], []
    with open(path) as f:
        for line in f:
            t, s = line.rstrip("\n").split("\t")[:2]
            times.append(float(t))
            sizes.append(int(float(s)))
    times = np.asarray(times) - float(times[0])
    feat, _, slot = fun(times, np.asarray(sizes))
    return np.array(feat, dtype=np.float64), float(slot)


if "groups" not in globals():
    print("重算各域 CIF（约 1~2 分钟）...")
    _pre = np.load(PRE_NPY, allow_pickle=True).item()
    groups = {}
    _x, _s, _y = [], [], []
    for _site in SITES:
        _sel = np.flatnonzero(_pre["label"] == _site)[:N_TRACE]
        _x.append(_pre["dataset"][_sel].astype(np.float64))
        _s.append(_pre["slot_duration"][_sel].astype(np.float64))
        _y.append(np.full(len(_sel), _site))
    groups["D1"] = dict(x=np.concatenate(_x), slot=np.concatenate(_s),
                        site=np.concatenate(_y))
    for _name, _dir in TRACE_DIRS.items():
        _x, _s, _y = [], [], []
        for _site in SITES:
            for _i in range(N_TRACE):
                _f, _sl = trace_features(f"{_dir}/{_site}-{_i}")
                _x.append(_f); _s.append(_sl); _y.append(_site)
        groups[_name] = dict(x=np.array(_x), slot=np.array(_s), site=np.array(_y))
        print(f"  {_name}: {len(_x)} 条")

DOMAINS = ["D1"] + list(TRACE_DIRS)


def to_db(a):
    return 20 * np.log10(np.clip(a, 1e-9, None))


# ---- 物理频率轴：每条 trace 先算自己的 rfft，再插值到公共物理频率栅格 ----
# slot_duration 被裁到 <= 0.08，所以各域的奈奎斯特下界恒为 6.25Hz
NYQ = min(1.0 / (2.0 * g["slot"].max()) for g in groups.values())
F_GRID = np.linspace(0.0, NYQ, 1000)
BAND_IDX = [(nm,
             int(np.searchsorted(F_GRID, lo)),
             len(F_GRID) - 1 if hi is None else int(np.searchsorted(F_GRID, hi)))
            for nm, lo, hi in BANDS]

if "spec_phys" not in globals():
    spec_phys = {}
    for _k, _g in groups.items():
        _x, _slot = _g["x"], _g["slot"]
        _out = np.empty((len(_x), 2, len(F_GRID)))
        for _i in range(len(_x)):
            _fr = np.fft.rfftfreq(_x.shape[-1], d=_slot[_i])
            _amp = np.abs(np.fft.rfft(_x[_i], axis=-1)) / _x.shape[-1]
            for _ch in range(2):
                _out[_i, _ch] = np.interp(F_GRID, _fr, _amp[_ch])
        spec_phys[_k] = _out

# ---- 归一化槽索引轴：直接对 1000 槽序列做 rfft，bin k = "每 1000 槽 k 个周期" ----
if "spec_slot" not in globals():
    spec_slot = {k: np.abs(np.fft.rfft(g["x"], axis=-1)) / g["x"].shape[-1]
                 for k, g in groups.items()}

# ---- 4 套表示，供「分析二」比较判别力 ----
REPS = [
    ("raw_CIF",      "时域基线：CIF 直接展平"),
    ("phys_dB",      "物理频率轴 dB 谱（含 DC、未归一化）= 天真做法"),
    ("phys_norm_dB", "物理频率轴 dB 谱（去 DC + 逐条能量归一化）"),
    ("slot_norm_dB", "归一化槽轴 dB 谱（去 DC + 逐条能量归一化）"),
]


def feats_of(rep, dom):
    g = groups[dom]
    n = len(g["x"])
    if rep == "raw_CIF":
        return g["x"].reshape(n, -1)
    if rep == "phys_dB":
        return to_db(spec_phys[dom]).reshape(n, -1)
    if rep == "phys_norm_dB":
        s = spec_phys[dom]
        s = s / np.clip(s.sum(axis=(1, 2), keepdims=True), 1e-12, None)
        return to_db(s[:, :, 1:]).reshape(n, -1)
    if rep == "slot_norm_dB":
        s = spec_slot[dom]
        s = s / np.clip(s.sum(axis=(1, 2), keepdims=True), 1e-12, None)
        return to_db(s[:, :, 1:]).reshape(n, -1)
    raise KeyError(rep)


for _k in DOMAINS:
    _g = groups[_k]
    print(f"{_k}: {len(_g['x']):5d} 条, slot_duration {_g['slot'].min():.4f}~{_g['slot'].max():.4f}s, "
          f"DC 谱均值(=总包数/1000) {spec_phys[_k][:, :, 0].mean():.3f}")

print(f"\n公共物理频率栅格 0 ~ {NYQ:.2f} Hz（{len(F_GRID)} 点）")
print("频段:", ", ".join("%s = idx %d..%d" % (nm, a, b) for nm, a, b in BAND_IDX))
print("\n4 套表示:")
for _rep, _desc in REPS:
    print(f"  {_rep:<14} {_desc}")
```

### 一、偏移的全局性 / 逐站点性

读法：

- **点** = 一个网站（`SITES` 里的 10 个）在该频段的平均偏移量，`跨站点 std` 就是这些点的散布
- **灰色条** = 域内 std（目标域与 D1 两边逐条 trace 的 dB 标准差取平均），代表"自然波动有多大"
- **竖直短横** = 这 10 个点的均值，也就是"全局偏移"这个分量
- 右侧 `x.xx` = `跨站点 std / 域内 std`，也就是判据里的那个比值


```python
# =====================================================================
# 分析一：D2/D6/D7 相对 D1 的频段偏移 —— 全局常数 还是 逐站点？
# =====================================================================
site_arr = np.array(SITES)


def site_mean_db(spec, site):
    # 每个网站一条平均 dB 谱
    return np.stack([to_db(spec[site == s]).mean(axis=0) for s in site_arr])


def within_std_db(spec, site):
    # 网站内逐 trace 的 dB 标准差，再对网站取平均
    return np.mean([to_db(spec[site == s]).std(axis=0) for s in site_arr], axis=0)


mean_D1 = site_mean_db(spec_phys["D1"], groups["D1"]["site"])
std_D1  = within_std_db(spec_phys["D1"], groups["D1"]["site"])

offsets, within = {}, {}
for tgt in TRACE_DIRS:
    m = site_mean_db(spec_phys[tgt], groups[tgt]["site"])
    w = within_std_db(spec_phys[tgt], groups[tgt]["site"])
    offsets[tgt] = m - mean_D1          # 逐站点偏移 (n_site, 2, n_freq)
    within[tgt]  = 0.5 * (w + std_D1)   # 两边域内 std 的平均

hdr = "%-4s%-6s%-14s%10s%11s%9s%7s   %s" % ("域", "通道", "频段", "全局偏移",
                                            "跨站点std", "域内std", "比值", "判断")
print(hdr)
print("-" * 78)
for tgt in TRACE_DIRS:
    for ch, chn in enumerate(["up", "down"]):
        for nm, a, b in BAND_IDX:
            o = offsets[tgt][:, ch, a:b + 1].mean(axis=1)   # 每个网站一个偏移量
            w = within[tgt][ch, a:b + 1].mean()
            ratio = o.std() / w
            v = "全局可校正" if ratio < 0.5 else ("部分逐站点" if ratio < 1.0 else "逐站点为主")
            print("%-4s%-6s%-14s%+9.1fdB%10.2f%9.2f%7.2f   %s"
                  % (tgt, chn, nm, o.mean(), o.std(), w, ratio, v))
    print()

# ---------------- 图 ----------------
lo_lim = min(-6.5, min(offsets[t][:, ch, a:b + 1].mean(axis=1).min()
                       for t in TRACE_DIRS for ch in range(2) for nm, a, b in BAND_IDX))
hi_lim = max(6.5, max(offsets[t][:, ch, a:b + 1].mean(axis=1).max()
                      for t in TRACE_DIRS for ch in range(2) for nm, a, b in BAND_IDX))
span = hi_lim - lo_lim

fig, axes = plt.subplots(2, 3, figsize=(13.5, 6.0))
for ci, tgt in enumerate(TRACE_DIRS):
    for ch in range(2):
        ax = axes[ch, ci]
        for bi, (nm, a, b) in enumerate(BAND_IDX):
            o = offsets[tgt][:, ch, a:b + 1].mean(axis=1)
            w = within[tgt][ch, a:b + 1].mean()
            ax.barh(bi, 2 * w, left=-w, height=0.60, color=GRIDC, zorder=1)
            ax.scatter(o, bi + np.linspace(-0.13, 0.13, len(o)), s=36, color=S1,
                       edgecolor=SURFACE, linewidth=0.9, zorder=3)
            ax.scatter([o.mean()], [bi], marker="|", s=120, color=INK,
                       linewidths=1.6, zorder=4)
            ax.annotate("%.2f" % (o.std() / w), xy=(0.985, bi),
                        xycoords=("axes fraction", "data"), ha="right", va="center",
                        color=INK_2, fontsize=8)
        ax.axvline(0.0, color=INK_2, lw=1.0, zorder=2)
        ax.set_xlim(lo_lim, hi_lim + 0.34 * span)     # 右侧留白给比值标注
        ax.set_ylim(-0.6, len(BAND_IDX) - 0.4)
        ax.set_yticks(range(len(BAND_IDX)))
        ax.set_yticklabels([nm for nm, _, _ in BAND_IDX])
        ax.set_title("%s vs D1   (%s)" % (tgt, "uplink" if ch == 0 else "downlink"))
        ax.grid(axis="x", color=GRIDC, lw=0.8)
        ax.set_axisbelow(True)

fig.supxlabel("band offset vs D1 (dB)      dot = one website   |   gray bar = within-domain std"
              "   |   vertical tick = mean   |   right number = std_across_sites / std_within",
              fontsize=9, color=INK_2)
fig.tight_layout(rect=[0, 0.055, 1, 1])
plt.show()
```

### 二、频域到底有没有判别力？

上面只量了**域偏移**。一个表示可以既域不变又毫无判别力（常数最极端），所以还得量**类可分性**。
这里用 **Fisher 比**：

$$\mathrm{Fisher} = \frac{\text{类间距离均方}}{\text{类内距离均方}}$$

（类 = 网站；距离是特征空间里的平方欧氏距离。）

对每套表示算两个数：

| 记号 | 怎么算 | 含义 |
|---|---|---|
| `同域 F(D1)` | 只用 D1 的样本，类 = 网站 | 这套表示在源域里的**判别力上限** |
| `跨域 F` | 把 D1 与目标域同样网站的样本**混成一类**，再算 | 域偏移被算进类内散度后，还剩多少可分性 |
| **保留率** | `跨域 F / 同域 F` | **接近 1 = 扛得住域偏移**；接近 0 = 域偏移把类结构抹平了 |

四套表示放在一起比较，正好回答两个问题：

- `phys_dB` → `phys_norm_dB`：**DC / 整体尺度**是不是域偏移的元凶？（前者是天真做法，含 DC 且不归一化）
- `phys_norm_dB` → `slot_norm_dB`：**频率轴选哪个**？物理 Hz 轴会把不同 `slot_duration` 的 trace
  放到不同的结构尺度上比较；归一化槽轴（bin k = 每 1000 槽 k 个周期）才是 CIF 本身所在的空间。

注意 `同域 F` 的**绝对值**在不同表示之间没有可比性（量纲不同），要比的是**保留率**。


```python
# =====================================================================
# 分析二：判别力 —— 类间/类内距离比（Fisher 比）及其跨域保留率
# =====================================================================
def fisher(F, site):
    # Fisher 比 = 类间距离均方 / 类内距离均方
    F = np.asarray(F, dtype=np.float64)
    site = np.asarray(site)
    uniq = np.unique(site)
    mus = np.stack([F[site == s].mean(axis=0) for s in uniq])
    W = np.mean([np.mean(np.sum((F[site == s] - m) ** 2, axis=1))
                 for s, m in zip(uniq, mus)])
    B = np.mean([np.sum((mus[i] - mus[j]) ** 2)
                 for i in range(len(uniq)) for j in range(i + 1, len(uniq))])
    return B / W


f_in, f_cross = {}, {}
for rep, _desc in REPS:
    s1 = groups["D1"]["site"]
    F1 = feats_of(rep, "D1")
    f_in[rep] = fisher(F1, s1)
    for tgt in TRACE_DIRS:
        F2 = feats_of(rep, tgt)
        f_cross[(rep, tgt)] = fisher(np.vstack([F1, F2]),
                                     np.concatenate([s1, groups[tgt]["site"]]))

print("Fisher 比 = 类间距离均方 / 类内距离均方（越大越可分）")
print("保留率 = 跨域 F / 同域 F，接近 1 说明该表示扛得住域偏移\n")
print("%-16s%12s" % ("表示", "同域F(D1)")
      + "".join("%12s%12s" % (t + " 跨域F", t + " 保留率") for t in TRACE_DIRS))
print("-" * 84)
for rep, _desc in REPS:
    line = "%-16s%12.2f" % (rep, f_in[rep])
    for tgt in TRACE_DIRS:
        line += "%12.2f%12.3f" % (f_cross[(rep, tgt)], f_cross[(rep, tgt)] / f_in[rep])
    print(line)

# ---------------- 图 ----------------
fig, ax = plt.subplots(figsize=(9.8, 4.0))
n_rep = len(REPS)
h = 0.24
maxv = 0.0
for di, (tgt, col) in enumerate(zip(TRACE_DIRS, [S1, S2, S3])):
    vals = [f_cross[(rep, tgt)] / f_in[rep] for rep, _ in REPS]
    maxv = max(maxv, max(vals))
    y = np.arange(n_rep) + (di - 1) * (h + 0.03)
    ax.barh(y, vals, height=h, color=col, label=tgt)
    for yy, v in zip(y, vals):
        ax.text(v + 0.012 * max(1.0, maxv), yy, "%.2f" % v, va="center", ha="left",
                color=INK_2, fontsize=8)

ax.axvline(1.0, color=INK_2, lw=1.0, ls="--")
ax.set_xlim(0, maxv * 1.22)
ax.set_yticks(np.arange(n_rep))
ax.set_yticklabels([r for r, _ in REPS])
ax.set_ylim(n_rep - 0.5, -0.5)          # 手工定 y 范围，下面的标注才不会飘到图例上
ax.annotate("no loss (1.0)", xy=(1.0, -0.40), xytext=(4, 0),
            textcoords="offset points", color=MUTED, fontsize=8, va="center")
ax.set_xlabel("Fisher ratio retained after mixing D1 with the target domain")
ax.set_title("Does the representation survive the domain shift?")
ax.grid(axis="x", color=GRIDC, lw=0.8)
ax.set_axisbelow(True)
ax.legend(loc="lower right", labelcolor=INK_2)
fig.tight_layout()
plt.show()
```

### 结论

#### 一、偏移本身是什么性质

| 域 | `跨站点std / 域内std` | 判断 | 全局偏移（10 站均值） |
|---|---|---|---|
| **D2** | 0.16 – 0.42 | **全局可校正** —— 10 个网站的偏移几乎重合 | 低频 −3.4 dB，中高频 ≈0 |
| **D6** | 0.63 – 1.53 | 混合：低频段逐站点，中高频偏全局 | 各段 +1.3 ~ +2.7 dB |
| **D7** | 1.03 – 1.84 | **逐站点为主** —— 偏移幅度与网站内容纠缠在一起 | 下行中/高频 **+6.5 ~ +6.7 dB**，DC 涨到 2.4 倍 |

**"这个域整体偏了多少"在 D2 上是个合法的问题，在 D7 上不是。** D7 的偏移不是一层能整体剥掉的增益，
而是被逐网站重塑过的 —— 这和它 DC（总包数）涨 2.4 倍是一致的。

> D6 下行低频段有 1~2 个网站的偏移达到 **+13 ~ +26 dB**，是明显离群点，把该段的跨站点 std 抬了上去。
> 该段的"逐站点为主"要打折看。

#### 二、频域有没有判别力 —— 结论和我先前的判断相反

| 表示 | 同域 F(D1) | D2 跨域F | D2 保留率 | D6 跨域F | D6 保留率 | D7 跨域F | D7 保留率 |
|---|---|---|---|---|---|---|---|
| `raw_CIF` | 1.03 | 0.56 | 0.55 | 0.73 | 0.71 | 0.15 | 0.14 |
| `phys_dB` | 3.00 | 2.65 | 0.88 | 1.45 | 0.48 | 0.48 | 0.16 |
| `phys_norm_dB` | 1.82 | 1.49 | 0.82 | 1.48 | 0.81 | 0.31 | 0.17 |
| `slot_norm_dB` | 1.43 | 1.32 | **0.92** | 1.18 | **0.83** | 0.26 | **0.18** |

**`slot_norm_dB`（归一化槽轴 + 去 DC + 能量归一化的 dB 幅度谱）在三个域上保留率都最高**，
而且跨域 Fisher 的**绝对值**也比 `raw_CIF` 高（D2：1.32 vs 0.56；D6：1.18 vs 0.73）。
所以在这个数据上，**频域幅度谱确实比原始 CIF 更扛域偏移** ——
原先"幅度谱丢掉了相位就没判别力"的判断，至少在这套 Fisher 口径下站不住。

三个可操作的细节：

1. **DC / 整体尺度是最脏的一维。** `phys_dB` 在 D6 上保留率只有 0.48，一去 DC 加能量归一化就跳到 0.81。
   而它的同域 F 高达 3.00 —— 说明它"很能分"，但分出来的相当一部分是**域**，不是网站。
2. **频率轴要用归一化槽轴，不要用物理 Hz 轴。** `phys_norm_dB → slot_norm_dB` 在 D2 上 0.82 → 0.92、
   D6 上 0.81 → 0.83，方向一致地变好。物理 Hz 轴会把 `slot_duration` 不同的 trace 放到**不同的结构尺度**上比较。
3. **D7 谁都救不了**（0.14 / 0.16 / 0.17 / 0.18），和第一部分的结论对上了：那个域的偏移是逐站点的。

#### 三、这个实验**不能**证明的东西

`raw_CIF` 是**未归一化**的原始包计数，而三套频域表示都做了 **dB（对数压缩）+ 能量归一化**。
所以"频域更好"里混着两部分的贡献：

- 傅里叶变换本身（把序列变成谱）
- log 压缩 + 能量归一化（拉平量纲、去掉整体尺度）

要干净地归因，缺一个控制组：**对 raw CIF 做同样的 log 压缩 + 能量归一化，但不做傅里叶变换**。
如果那个控制组的保留率也接近 `slot_norm_dB`，那真正起作用的就是归一化而不是"频域"。

另外，这里量的是**特征空间里的 Fisher 比**，是线性可分性的代理指标，不是分类准确率。
方向可信，但要落到论文表里的 N-shot 精度数字，还得跑 `fineTuning-AE.py` 那套线性探针。
